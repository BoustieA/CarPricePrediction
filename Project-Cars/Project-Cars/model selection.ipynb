```python
cat_data=df.dtypes.loc[df.dtypes=="object"].index
num_data=df.dtypes.loc[df.dtypes!="object"].index[1:]
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import Ridge, LinearRegression
from sklearn.preprocessing import FunctionTransformer
import xgboost as xgb

from sklearn.linear_model import LinearRegression, LogisticRegressionCV
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.neural_network import MLPRegressor
from sklearn.ensemble import GradientBoostingRegressor, HistGradientBoostingRegressor,RandomTreesEmbedding
from sklearn.model_selection import GridSearchCV
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
```

```python

C1=ColumnTransformer([("prix",'passthrough',["Prix"]),
                      ("ohe", OneHotEncoder(handle_unknown="ignore",sparse_output=False), cat_data),
                      ("num",StandardScaler(),num_data)])


P=Pipeline([("Transform", C1)])#,("P",C2)])
```

```python
train,test=train_test_split(df,train_size=0.8,random_state=42)
```

```python
T=C1.fit_transform(train)#test preprocessing
```

```python
P.fit(train)
```

```python
df_train=P.transform(train)
df_test=P.transform(test)
```

```python
df_train.shape
```

```python
X_train=df_train[:,1:]
X_test=df_test[:,1:]
Y_train=df_train[:,0].ravel()
Y_test=df_test[:,0].ravel()
```

```python
X_train.shape,Y_train.shape,X_test.shape,Y_test.shape
```

# Comparaison de modèle avec hyperparamètre par défaut

```python
from sklearn.model_selection import cross_val_score
S={}
for estimator in [LinearRegression(),Ridge(),KNeighborsRegressor(),RandomForestRegressor(),XGBRegressor(),LGBMRegressor()]:
    score=cross_val_score(estimator,X_train,Y_train,scoring='r2',cv=5)
    S[estimator]=[np.mean(score),np.std(score)]
S
```

```python
param_grid={"max_depth":[5,10,None],"min_samples_leaf":[1,2,3],"warm_start":[True,False],"min_samples_split":[2,3,4]}

GCV=GridSearchCV(RandomForestRegressor(),param_grid=param_grid,scoring='r2')
GCV.fit(X_train,Y_train)
GCV.best_params_, GCV.best_score_
```

```python
param_grid=params = {
    "max_depth":[3,4,5],       # Tree depth
    "learning_rate":[0.1,0.01], # Learning rate
    "n_estimators":[1000,10000],  # Number of trees
    "reg_alpha":[1,10],     # L1 regularization
    "reg_lambda":[1,10],    # L2 regularization
    "gamma":[0,5,10]          # Min loss reduction for splits
}


GCV=GridSearchCV(XGBRegressor(),param_grid=param_grid,scoring='r2')
GCV.fit(X_train,Y_train)
GCV.best_params_,GCV.best_score_
```
