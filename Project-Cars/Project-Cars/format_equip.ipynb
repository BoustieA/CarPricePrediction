```python
import pandas as pd
df=pd.read_csv("Data/occasion_avec_prix_neuf_filled.csv").drop(columns="Unnamed: 0")
```

```python
#get unique values
import ast
import numpy as np
O=df.options.values
O=[ast.literal_eval(o) for o in O]
L=[]
for i in O:
    for j in i:
        L+=[j]
L=np.unique(L)
L
```

```python
#create column for each option
for option in L:
    df.loc[:,option]=df.loc[:,"options"].apply(lambda x: option in x)
```

```python
df=df.drop(columns="options")
```

```python
df.isnull().sum()
```

```python
df.to_csv("Data/car_fused_filled.csv")
```
