```python
sns.histplot(data=df,x="Energie")
plt.xticks(rotation=45)
plt.show()
plt.tight_layout()
```

```python
sns.histplot(data=df,x="Marque")
plt.xticks(rotation=45)
plt.show()
plt.tight_layout()
```

```python
sns.histplot(data=df,x="Prix",hue="Marque",multiple="stack")
plt.xticks(rotation=45)
plt.title("prix vs marque")
plt.show()
```

```python
sns.scatterplot(data=df,x="Prix",y="Kilometre")
plt.xticks(rotation=45)
plt.show()
plt.tight_layout()
```

```python
sns.scatterplot(data=df, x="Prix", y="Chevaux")
plt.xticks(rotation=45)
plt.show()
plt.tight_layout()
```

```python
df.columns
```

```python
sns.histplot(data=df, x="Prix", hue="Boites de vitesse", multiple="stack")
plt.xticks(rotation=45)
plt.show()
plt.tight_layout()
```

```python
sns.histplot(data=df, x="Prix", hue="Energie", multiple="stack")
plt.xticks(rotation=45)
plt.tight_layout()
plt.xlim(0,100000)
plt.show()
```

```python
sns.histplot(data=df, x="Prix", hue="Annee", multiple="stack")
plt.xticks(rotation=45)
plt.tight_layout()
plt.xlim(0,100000)
plt.show()
```

```python
plt.figure(figsize=(8,8))
sns.histplot(data=df, x="Prix", hue="type", multiple="stack")
plt.xticks(rotation=45)
plt.show()
plt.tight_layout()
```
